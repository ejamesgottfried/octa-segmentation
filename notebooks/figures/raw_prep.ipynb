# Raw vs. Preprocessed Figure

Single side-by-side figure (raw ROSE-1 SVC test image vs. its CLAHE+median
preprocessed version) saved for the writeup. See `preprocessing/ROSE_preprocess.ipynb`
for the pipeline that produced the preprocessed copy.


```python
import glob
import cv2
import matplotlib.pyplot as plt

prep_root = '/users/egottfri/code/octa-segmentation/data/preprocessed/SVC'
raw_root = '/files22_lrsresearch/ENG_Lee-Lab_Shared/group/data/public/rose_dataset/ROSE-1/SVC'

prep_imgs = sorted(glob.glob(f'{prep_root}/test/img/*'))
raw_imgs = sorted(glob.glob(f'{raw_root}/test/img/*'))

fig, axes = plt.subplots(1, 2, figsize=(8, 3.5), gridspec_kw={'wspace': 0.02})

img = cv2.imread(raw_imgs[0], cv2.IMREAD_GRAYSCALE)
axes[0].imshow(img, cmap='gray')
axes[0].set_title('Raw', fontsize=10)

img = cv2.imread(prep_imgs[0], cv2.IMREAD_GRAYSCALE)
axes[1].imshow(img, cmap='gray')
axes[1].set_title('Preprocessed', fontsize=10)

for ax in axes:
    ax.set_xticks([])
    ax.set_yticks([])

plt.savefig('../figures/raw_vs_preprocessed_examples.png', dpi=300, bbox_inches='tight')
plt.show()
```
